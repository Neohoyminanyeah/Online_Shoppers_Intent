```python
# pair plot

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from ucimlrepo import fetch_ucirepo

shoppers = fetch_ucirepo(id=468) # fetching dataset

X = shoppers.data.features #DF: should be 17 feature columns
y = shoppers.data.targets #DF: the revenue column

df = pd.concat([X, y], axis=1) #combines to one df

print(df.shape)            
print(df.columns.tolist())  

cols = ["ProductRelated", "ProductRelated_Duration",
        "BounceRates", "ExitRates", "PageValues", "Revenue"]

sns.pairplot(
    df[cols].sample(3000, random_state=42),
    hue="Revenue",
    corner=True,
    plot_kws={"alpha": 0.4, "s": 15},
    diag_kind="kde",
)
plt.show()
```
